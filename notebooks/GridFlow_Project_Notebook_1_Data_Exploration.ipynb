# NOTES


**Data Exploration and Time Handling**


- Loaded the NESO Historic Demand Data for 2025. This included 17520 30min observations


*   Confirmed that there were no missing values in the raw dataset
*   Reviewed the Key NESO Concepts (website):
    - National Demand (ND): GB metered genreation requirement. Excludes station load, pumped storage pumping and interconnector exports.
    - Transmission System Demand (TSD): ND plus the additional requirements including station load, pumped storage pumping and interconnector exports
    - Embedded win/solar: Estiamted distribution connected generation, not fully visible through transmission level metering.
* Confirmed Settelment Period 1 is equal to 00:00 - 00:30, meaning 48 30min periods in one 24 hr day.

* Identified daylight saving (DST) exceptions:
    - 30/03/25 = 46 settlement periods (spring)
    - 26/10/25 = 50 settlement periods (autumn)

* Naive local time would have been unsuitable because DST can create repeated local times

* Created a continuous UTC timefram (MODEL_TIME) that progresses in 30min intervals.

* Retained SETTLEMENT_DATE and SETTLEMENT_PERIOD

* Verified that the UTC timeline is continuous on the days mentioned above
