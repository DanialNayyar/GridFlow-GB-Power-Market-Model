```python
cost = [technology_parameters["WIND"]["marginal_cost"],
     technology_parameters["SOLAR"]["marginal_cost"],
     technology_parameters["NUCLEAR"]["marginal_cost"],
     technology_parameters["BIOMASS"]["marginal_cost"],
     technology_parameters["GAS"]["marginal_cost"]]


def dispatch_func(data_df):
  results_list = []
  A_coeffs = np.array([[1,1,1,1,1]])

  for row in range(len(data_df)):
    bounds_row = [(0,data_df["TOTAL_WIND"].iloc[row]),
      (0,data_df["SOLAR"].iloc[row]),
      (0,technology_parameters["NUCLEAR"]["capacity"]),
      (0,technology_parameters["BIOMASS"]["capacity"]),
      (0,technology_parameters["GAS"]["capacity"])]

    b_coeffs_row = np.array([data_df["DISPATCH_DEMAND"].iloc[row]])

    result_i = linprog(c = cost,
                  A_eq=A_coeffs,
                  b_eq=b_coeffs_row,
                  bounds = bounds_row,
                  method='highs')

    if result_i.success:
      results_dict_i = {
          "Hourly Operating Cost":result_i.fun,
          "Operating Cost (30mins)":result_i.fun*0.5,
          "Optimal Wind":result_i.x[0],
          "Optimal Solar":result_i.x[1],
          "Optimal Nuclear":result_i.x[2],
          "Optimal Biomass":result_i.x[3],
          "Optimal Gas":result_i.x[4]
      }

      results_list.append(results_dict_i)


  return results_list



```

```python
results_list = dispatch_func(data_df)
```

# Baseline Economic Dispatch Model - A Summary

The basleine mode formulates electricity dispatch as a simplified least cost ecnomic dispatch problem across the 17520 30min periods in 2025. The optimsised technologies include, wind, solar, nuclear and biomass whereas hydro, coal, imports and others were treated as a fixed observed supply. Wind and solar use their historical 30min output as time-varing availability while nuclear, gas and biomass used their fixed capacity limits based on the observed maximums over 2025.

The optimisation uses simplified short run marginal costs as listed in the table below:

$$
\begin{array}{|c|c|c|}
\hline
\text{Technolgy} & \text{SRMC} & \text{Source} \\
\hline
Wind  & £ 04.00/MWh & Source 2 \\
Solar & £ 00.00/MWh & Source 2 \\
Nuclear & £ 17.00/MWh & Source 3 \\
Gas & £ 80.00/MWh & Source 1 \\
Biomass & £ 85.00/MWh & Source 1 \\
\hline
\end{array}
$$



The model represents a simplified merit-order dispatch rather than unit commitment or full market operation.

The demand that has to be served by the optimised technologies is determined through:

$$
DISPATCH\_ DEMAND = UNDERLYING\_DEMAND - HYDRO-COAL-IMPORTS-OTHER
$$

Therefore the optimiser was responsible for meeting the portion of the underlying demand not already supplied by the "fixed" generation technologies.

The annual dispatch requirement was around 213 TWh. A final reconstruction demand (carried out as a final consistency check) produced the same value.


The resulting baseline dispatch was:
- Wind Energy Modelled: 85.844437 TWh
- Solar Energy Modelled: 18.660748 TWh
- Nuclear Energy Modelled: 44.5080475 TWh
- Biomass Energy Modelled: 0.011929 TWh
- Gas Energy Modelled: 63.8542245 TWh
- TOTAL MODELLED: 212.879386 TWh

The modelled variable operating cost was approaximately £6.21 billion

The periods each technology was dispatched is below:
- Number of Biomass Dispatch Periods: 38
- Number of Gas Dispatch Periods: 15049
- Number of Wind Dispatch Periods: 17520
- Number of Solar Dispatch Periods: 8934



The optimier passed the validation checks, producing one result for each settlement period, all techs remained within the specified availability/capacity bounds and the optimised generation matched the dispatch demand for every 30min period.

Baseline renewable curtailment was negligebale with only 1.2 GWh of wind being curtalied across 6 30min periods, while no solar was curtailed. This suggests that the 2025 baseline could accomadte almost all of the renewable availabilty represente by historical wind and solar technologies.

When comparing with the historical generation the behaviour of the pure SRMC based dispatch, wind remaind almost unchanged and solar was unchanged. However, modelled unclear generaion increased by approximately 10.4 TWh, whilst biomass fell by approximately 19.9 TWh and gas by 13.5 TWh. This is ofcourse consistent with the simplified merid order. The optimiser prefers the lower cost nuclear before gas and uses the more expensive biomass if the cheaper available supply is insufficient.

It is important to note that the baseline should not be interepreted as an attempt to reproduce the actual GB dispatch. It does not include many factors such as minimum stable generation, startup costs, outages, network constraint as well as the hydro and import economics. Conseuqently the baseline provides a transparent counterfactual least-cost benchmark against which the later renewable and battery scenarios can be compared.

```python
def dispatch_func(data_df):
  results_list = []
  A_coeffs = np.array([[1,1,1,1,1]])

  for row in range(len(data_df)):
    bounds_row = [(0,data_df["TOTAL_WIND"].iloc[row]),
      (0,data_df["SOLAR"].iloc[row]),
      (0,technology_parameters["NUCLEAR"]["capacity"]),
      (0,technology_parameters["BIOMASS"]["capacity"]),
      (0,technology_parameters["GAS"]["capacity"])]

    b_coeffs_row = np.array([data_df["DISPATCH_DEMAND"].iloc[row]])

    result_i = linprog(c = cost,
                  A_eq=A_coeffs,
                  b_eq=b_coeffs_row,
                  bounds = bounds_row,
                  method='highs')

    if result_i.success:
      results_dict_i = {
          "Hourly Operating Cost":result_i.fun,
          "Operating Cost (30mins)":result_i.fun*0.5,
          "Optimal Wind":result_i.x[0],
          "Optimal Solar":result_i.x[1],
          "Optimal Nuclear":result_i.x[2],
          "Optimal Biomass":result_i.x[3],
          "Optimal Gas":result_i.x[4],
          "Marginal Price":result_i.eqlin.marginals[0]
      }

      results_list.append(results_dict_i)


  return results_list

```
